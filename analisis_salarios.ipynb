```python
#1. El dataset cargado en Python: 

#importacion de mi dataset desde kaggle
path = kagglehub.dataset_download("nalisha/job-salary-prediction-dataset")

print("Ruta del dataset:", path)

CSV_salarios = os.listdir(path)
print(CSV_salarios)
```

```python
#2. Un .head() del dataset:

#importacion del CSV para trabajar directamente con mi dataframe: Job Salary Prediction
archivo = CSV_salarios[0]
ruta_dataset = path + "/" + archivo
df = pd.read_csv(ruta_dataset)

df.head(11) 
```
