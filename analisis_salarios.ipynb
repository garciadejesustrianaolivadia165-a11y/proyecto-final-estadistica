4. Análisis de variables numéricas

```python
# 4. Análisis de variables numéricas 
#- `.describe()`
 
num = df.select_dtypes(include=['int64','float64']) # aqui seleciono solamente cuales son mis variables numericas
print(f"\n Variables encontras: {list(num.columns)}\n")
print("estadisticas descripctivas: ")
num.describe().round(2)
```

```python
# 4. Análisis de variables numéricas 
#- Histogramas (`histplot`) 

fig, axes = plt.subplots(2, 2, figsize=(14, 10)) #esto es para hacer el fondo de los histogramas
fig.suptitle('Histogramas de las variables numericas', fontsize=15, fontweight='bold') #titulo del histograma

cols_num = num.columns.tolist()

for i, col in enumerate(cols_num): 
    ax = axes[i//2][i%2]
    sns.histplot(data=df, x = col, bins=30, kde=True, color='blue', ax=ax)
    ax.set_title(f'Distribucion de {col}', fontsize=9)
    ax.set_xlabel(col)
    ax.set_ylabel('Frecuencia')
    ax.grid(axis='y', alpha=0.3)


print("Histogramas readys")
```

```python
# 4. Análisis de variables numéricas 
#- Boxplots  

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Boxplots de las variables numericas', fontsize=15, fontweight='bold')

for i,col in enumerate(cols_num):
    ax = axes[i//2][i%2]
    sns.boxenplot(data=df, y=col, color='blue', ax=ax)
    ax.set_title(f'Boxplot de {col}', fontsize=9)
    ax.set_label(col)
    ax.grid(axis='y', alpha=0.3)

print(" Boxplots readys")
```

### Interpretación — Variables Numéricas

**experience_years (Años de experiencia):**
- Va de 0 a 20 años, con media alrededor de 10 años.
- Distribución aproximadamente uniforme (sin sesgo marcado).
- No se observan outliers extremos.

**skills_count (Número de habilidades):**
- Va de 1 a 19, con media cercana a 10.
- Distribución bastante simétrica.

**certifications (Certificaciones):**
- Va de 0 a 5, con media de ~2.5.
- Distribución relativamente uniforme.

**salary (Salario):**
- Media alrededor de $145,000 con desviación estándar de ~$37,000.
- El boxplot muestra una distribución con cola derecha (algunos salarios muy altos).
- Sí existen outliers en la parte superior (salarios muy elevados).
- Esto indica sesgo positivo (hacia la derecha).

5. Análisis de variables categóricas

- `value_counts()`  

- Gráficos de barras (`countplot`)  



Interpretación requerida:

- ¿Qué categorías dominan?  

- ¿Hay desbalance en los datos?  



```python
#5. Análisis de variables categóricas
#- `value_counts()`  

categorias = df.select_dtypes(include=['object'])
print(f"\n Variables Categoricas {list(categorias.columns)}\n")

#calculo de value_counts de cada variable
for col in categorias.columns:
    print("{col.upper}: ")
    print(df[col].value_counts())
    print(f"categorias que son unicas (Total): {df[col].nunique()}")
    print("_" * 50)



```

```python
#5. Análisis de variables categóricas
#- Gráficos de barras (`countplot`)  

fig, axes = plt.subplots(3, 2, figsize=(16, 16))
fig.suptitle('Distribución de Variables Categóricas', fontsize=15, fontweight='bold')

cols_cat = categorias.columns.tolist()


for i, col in enumerate(cols_cat):
    ax = axes[i // 2][i % 2]
    orden = df[col].value_counts().index
    sns.countplot(data=df, x=col, order=orden, palette='viridis', ax=ax) # palette es uno de los parametros de la libreria seaborn, para colocar varios colores
    ax.set_title(f'Distribución de {col}', fontsize=9)
    ax.set_xlabel(col)
    ax.set_ylabel('Cantidad')
    ax.tick_params(axis='x', rotation=45, labelsize=8)
    ax.grid(axis='y', alpha=0.3)

plt.tight_layout() # esto es para que la letra no aparezca una arriba de la otra
print(" Graficos de barras redy")
    
```

### Interpretación — Variables Categóricas

**job_title (Título del trabajo):**

- El rol más frecuente es Backend Developer.

**education_level (Nivel educativo):**
- 5 categorías: High School, Diploma, Bachelor, Master, PhD.
- Master es el más común (~50,000 registros).
- No hay desbalance extremo entre niveles.

**industry (Industria):**

**location (Ubicación):**

6. Medidas de dispersión

Calcular:

- Rango  

- Varianza  

- Desviación estándar  



Interpretación requerida:

- ¿Los datos están concentrados o dispersos?  

- ¿Qué implica esto para el análisis?  

```python
# 6. Medidas de dispersión

for col in cols_num:
    datos = df[col].dropna()
    rango = datos.max() - datos.min()
    varianza = datos.var().round(2)
    desviacion = datos.std().round(2)
    coeficiente = (desviacion/datos.mean()*100).round(2)

    print(f"\n Metricas de {col.upper()}")
    print(f"   Mínimo:                  {datos.min():>12,.2f}")
    print(f"   Maximo:                  {datos.max():>12,.2f}")
    print(f"   Rango:                   {rango:>12,.2f}")
    print(f"   Varianza:                {varianza:>12,.2f}")
    print(f"   Desviacion estándar:     {desviacion:>12,.2f}")
    print(f"   Coef. de variacion (%):  {coeficiente:>12,.2f}%")
    print("_" * 45)

 
```

### Interpretación — Medidas de Dispersión



**¿Los datos están concentrados o dispersos?**

- **experience_years**: Rango de 0 a 20. Desviación estándar ~6, lo que indica
  una dispersión moderada. Los años de experiencia están distribuidos de forma
  bastante uniforme en todo el rango.

- **skills_count**: Rango de 1 a 19. Dispersión similar a experience_years.
  Los datos no se concentran en un valor específico.

- **certifications**: Rango de 0 a 5. Desviación estándar ~1.7,
  relativamente concentrados alrededor de la media (~2.5).

- **salary**: Es la variable con mayor dispersión en términos absolutos.
  Desviación estándar de ~$37,000 sobre una media de ~$145,000.
  Esto significa que los salarios varían considerablemente entre empleados,
  lo que hace que esta variable sea interesante para modelar y predecir.

**¿Qué implica esto para el análisis?**
La dispersión en el salario sugiere que existen factores significativos
que explican esas diferencias (experiencia, educación, industria, etc.).
Esto valida que un modelo de regresión tendría sentido para predecir el salario.
