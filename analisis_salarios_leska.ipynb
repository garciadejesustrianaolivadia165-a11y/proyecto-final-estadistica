```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar dataset con tu ruta
df = pd.read_csv(r"C:\Users\Waleska_AS\OneDrive\modulo de estadistica\job_salary_prediction_dataset.csv")

# Verificar que cargó bien
df.head()
```

```python
df.select_dtypes(include=['int64','float64']).skew()
```

### Análisis de sesgo (Skewness)

El sesgo permite analizar la distribución de los datos.

Se observa que el salario presenta un sesgo positivo, lo que indica la presencia de valores extremos altos. 
Las demás variables muestran una distribución más equilibrada.

```python
#3. RELACIÓN ENTRE VARIABLES
sns.scatterplot(data=df, x='experience_years', y='salary')
plt.title("Relación entre experiencia y salario")
plt.show()

df_num = df.select_dtypes(include=['int64','float64'])

sns.heatmap(df_num.corr(), annot=True, cmap='coolwarm')
plt.title("Matriz de correlación")
plt.show()
```

### Relación entre variables

Se observa una relación positiva entre la experiencia y el salario, lo que indica que a mayor experiencia, mayor ingreso.

La matriz de correlación confirma que variables como experiencia, habilidades y certificaciones influyen en el salario.
